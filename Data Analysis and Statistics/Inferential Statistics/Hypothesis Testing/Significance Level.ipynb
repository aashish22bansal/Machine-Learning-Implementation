# Significance Level

The **Significance Level**, denoted by the Greek letter alpha ($\alpha$), is the probability threshold you set for making a wrong decision. Specifically, it is the exact probability of rejecting the null hypothesis ($H_0$) when it is actually true—a mistake known as a **Type I Error** (or a false positive).

Think of the significance level as your "burden of proof." Before you ever run an experiment or calculate a test statistic, you must decide how much evidence you require to declare a result meaningful, and how much risk you are willing to take of being wrong.

### How It Works in Practice

When you run a statistical test, you calculate a p-value (the probability that your observed results were just random noise). You then compare that p-value directly to your significance level:

* **If p-value $\leq \alpha$:** The result is statistically significant. You reject the null hypothesis. The evidence met your burden of proof.
* **If p-value $> \alpha$:** The result is not statistically significant. You fail to reject the null hypothesis. The evidence was too weak.

### Choosing Your Alpha

You must set $\alpha$ *before* looking at your data to avoid bias. There is no mathematical law dictating what $\alpha$ must be; it is a human decision based entirely on how dangerous a false positive is in your specific context.

| Significance Level ($\alpha$) | Confidence Level ($1 - \alpha$) | When It Is Used | The Trade-off |
| --- | --- | --- | --- |
| **0.05 (5%)** | 95% | The universal standard for most academic research, business analytics, and A/B testing. | A balanced approach. You accept a 1 in 20 chance of a false positive. |
| **0.01 (1%)** | 99% | Medical research, aerospace engineering, or any field where a false positive is dangerous or highly expensive. | Very strict. It minimizes false positives but makes it much harder to detect a real effect (increasing the risk of a Type II error/false negative). |
| **0.10 (10%)** | 90% | Early-stage exploratory research, or scenarios where missing a potential effect is worse than a false alarm. | Lenient. It is easy to find "significant" results, but 1 in 10 of them will be purely random noise. |

### The Trade-off: Alpha vs. Beta

You might wonder: *If false positives are bad, why not always set $\alpha = 0.0001$?*

In statistics, there is a constant seesaw between Type I errors ($\alpha$, false positives) and Type II errors ($\beta$, false negatives). If you make your significance level incredibly strict to avoid a false positive, you simultaneously require massive amounts of evidence to prove *anything*.

If a pharmaceutical company sets $\alpha$ too low when testing a genuinely life-saving drug, the test might fail to reach that extreme threshold, resulting in a false negative. The drug gets scrapped, and the real effect is missed. Choosing your significance level is about finding the right balance of risk for your specific project.
