# Confidence Levels

A **Confidence Level** is the mathematical complement to your Significance Level ($\alpha$). If the significance level is the amount of risk you are willing to take of being wrong, the confidence level is how certain you want to be that your results are right.

Specifically, it represents the percentage of times that your statistical methodology would produce a correct result if you were to repeat your experiment or sampling process over and over again under the exact same conditions.

### The Math: Two Sides of the Same Coin

Confidence Levels and Significance Levels are directly tethered to each other so that they always add up to 1 (or 100%).

$$\text{Confidence Level} = 1 - \alpha$$

If you set your significance level ($\alpha$) at $0.05$ (meaning you accept a 5% chance of a false positive), your confidence level is automatically $0.95$, or **95%**.

### How It Connects to Confidence Intervals

While hypothesis testing usually gives you a definitive "yes/no" answer (reject or fail to reject the null hypothesis), confidence levels are most frequently used to build **Confidence Intervals**.

A confidence interval gives you a *range of values* where you believe the true population number actually lives, based on your sample data.

For example, if you run a poll of 1,000 voters and find that 54% support a candidate, you know that exactly 54% of the *entire country* probably doesn't support them. There is some margin of error. A 95% confidence interval might tell you: *"We are 95% confident that the true national support for this candidate is between 51% and 57%."*

### Standard Confidence Levels

Just like with $\alpha$, the confidence level you choose depends on the stakes of your decision:

| Confidence Level | Alpha ($\alpha$) | What it means practically | Common Use Cases |
| --- | --- | --- | --- |
| **90%** | $0.10$ | You are accepting more risk. The resulting confidence interval will be narrower (more precise), but there is a 1 in 10 chance the true value isn't actually in that range. | Directional business decisions, early exploratory data analysis, marketing trends. |
| **95%** | $0.05$ | The universal gold standard. It provides a solid balance between being precise and being accurate. | Academic research, A/B testing, standard data science modeling. |
| **99%** | $0.01$ | You are demanding extreme certainty. Your resulting confidence interval will be very wide (less precise) to ensure it captures the true value 99 times out of 100. | Medical trials, aviation safety, mission-critical engineering. |

### The "Repeated Experiment" Misconception

The most common mistake people make is interpreting a 95% confidence interval as a strict probability for a *single* event.

* **Wrong:** "There is a 95% chance that the true population mean is inside this specific interval." (Once the interval is calculated, the true mean is either inside it or it isn't—the probability is either 100% or 0%).
* **Right:** "If we repeated this exact sampling process 100 times, 95 of the intervals we calculate would contain the true population mean." The 95% refers to the reliability of your *process*, not the specific numbers you generated this one time.
