# One-tailed Test

A **one-tailed test** (also known as a directional test) is a hypothesis test where you are looking for an effect or difference in *one specific direction*.

Instead of asking, "Is there any difference at all?" you are asking a much more targeted question: "Is the new value strictly *greater than* the old value?" or "Is the new value strictly *less than* the old value?"

Because you are only looking in one direction, you place your entire Significance Level ($\alpha$) into just one tail of your probability distribution.

### The Two Types of One-Tailed Tests

The direction of your Alternative Hypothesis ($H_a$) dictates which tail of the distribution you focus on:

| Test Type | The Question Being Asked | Alternative Hypothesis ($H_a$) | Where the $\alpha$ Region Lives |
| --- | --- | --- | --- |
| **Right-Tailed Test** | Is the new metric significantly *greater* than the baseline? | $H_a > H_0$ | The entire $\alpha$ area is shaded on the far **right** side of the curve. |
| **Left-Tailed Test** | Is the new metric significantly *less* than the baseline? | $H_a < H_0$ | The entire $\alpha$ area is shaded on the far **left** side of the curve. |

---

### Why Use a One-Tailed Test?

The main advantage of a one-tailed test is that it gives you more **statistical power** in your chosen direction.

Because you aren't splitting your $\alpha$ threshold in half (like you do in a two-tailed test where $\alpha = 0.05$ becomes $0.025$ on each side), your critical value sits closer to the center of the distribution. This means **you do not need as large of an effect to achieve statistical significance**, as long as the effect moves in the direction you predicted.

**Example Scenario:**
Imagine you are testing a new manufacturing process designed specifically to *reduce* the defect rate.

* You don't care if the new process increases the defect rate; if it does, you simply won't adopt it.
* You only care if it successfully lowers defects.
* By using a left-tailed test, you focus all your statistical power on detecting a reduction, making it easier to prove the process works if it actually does.

### The Major Risk: Blindness to the Opposite Direction

The danger of a one-tailed test is that it makes you completely blind to effects in the un-tested direction.

If you run a right-tailed test looking for an increase in sales, but your new marketing campaign actually causes a catastrophic drop in sales, your one-tailed test will return a result of "not statistically significant." It treats massive failure exactly the same as "no effect at all."

Because of this risk, the scientific and statistical community has strict rules for using one-tailed tests:

1. You must decide to use it *before* you collect or look at any data. You cannot look at the data, see it trending upward, and then decide to run a right-tailed test to make it easier to get a low p-value.
2. You must have a very strong theoretical or practical justification for why a result in the opposite direction is either impossible or completely irrelevant to your decision-making.
