# Two-tailed Test

A **two-tailed test** (also known as a non-directional test) is a hypothesis test where you are looking for any statistically significant difference, regardless of the direction.

Instead of asking if a new metric is strictly greater than or less than a baseline, you are asking a much broader question: "Is the new value *different* from the old value?"

Because you are looking for a change in either direction, you must split your Significance Level ($\alpha$) perfectly in half, placing one rejection region in the far left tail of your probability distribution and the other in the far right tail.

### The Math: Splitting Alpha ($\alpha$)

In a two-tailed test, your Alternative Hypothesis ($H_a$) asserts that the observed value is simply not equal to ($\neq$) the null value.

If you set your overall significance level at $\alpha = 0.05$ (a 5% error rate), you cannot put that entire 5% on one side of the curve. You must divide it:

* **Left Tail Rejection Region:** $\alpha / 2 = 0.025$ (2.5%)
* **Right Tail Rejection Region:** $\alpha / 2 = 0.025$ (2.5%)

If your test statistic falls into *either* of these 2.5% regions, you reject the null hypothesis.

### Why Use a Two-Tailed Test?

The two-tailed test is the universal default in statistics, A/B testing, and scientific research because it is the safest and most objective approach.

**The primary advantage is that it prevents you from being blind to unexpected outcomes.**
If you roll out a new algorithm that you strongly believe will reduce error rates, a two-tailed test will confirm if it successfully reduces them. However, if there is a bug in the code and the new algorithm drastically *increases* error rates, a two-tailed test will also flag that unexpected catastrophic failure as statistically significant.

### The Trade-Off: A Higher Bar for Significance

Because you split your $\alpha$ in half, the critical values that mark the boundaries of your rejection regions are pushed further out from the center of the distribution.

This means **you need a larger, more extreme effect to achieve statistical significance** in a two-tailed test compared to a one-tailed test.

| Feature | Two-Tailed Test | One-Tailed Test |
| --- | --- | --- |
| **Direction** | Any difference ($\neq$) | Specific direction ($>$ or $<$) |
| **Alpha Placement** | Split in half across both tails ($\alpha/2$) | Concentrated entirely in one tail ($\alpha$) |
| **Risk** | Harder to prove a small effect is significant. | Completely blind to effects in the un-tested direction. |
| **Best Use Case** | When you want to be completely objective and catch any deviation from the baseline. | When a change in the opposite direction is practically impossible or irrelevant. |
