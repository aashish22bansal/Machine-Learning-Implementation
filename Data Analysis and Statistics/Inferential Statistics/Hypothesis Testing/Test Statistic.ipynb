# Test Statistic

A **test statistic** is a single, standardized number calculated from your sample data. It acts as a summary of the evidence against the null hypothesis ($H_0$).

Think of it as a signal-to-noise ratio. It compares the size of the effect or difference you observed in your data (the signal) to the natural, expected variance in the data (the noise).

If the test statistic is large enough, it means your signal has overpowered the noise, pushing your result past the critical value and into the rejection region.

## The Universal Formula

While there are many specific formulas for different types of tests, almost all test statistics follow this underlying logic:

$$\text{Test Statistic} = \frac{\text{Observed Data} - \text{Expected Data under } H_0}{\text{Standard Error of the Data}}$$

* **Observed Data:** The actual metric you calculated from your sample (e.g., your sample mean).
* **Expected Data:** The baseline value if the null hypothesis were perfectly true (often zero, or a historical average).
* **Standard Error:** A measure of how much random variance or "wobble" is expected in your data based on your sample size.

## Common Types of Test Statistics

The specific test statistic you use depends entirely on the structure of your data and the question you are trying to answer:

| Statistic | Used For | Example Scenario |
| --- | --- | --- |
| **Z-score ($z$)** | Comparing a sample mean to a population mean when the sample size is large (typically $>30$) and the population variance is known. | Testing if the average conversion rate of a website with 10,000 visitors differs from the historical average. |
| **T-score ($t$)** | Comparing means when the sample size is small or the population variance is unknown (very common in practice). | Testing the effect of a new fertilizer on the growth of 15 plants. |
| **F-value ($F$)** | Comparing the variances of two or more groups (used in ANOVA tests). | Determining if average salaries differ significantly across four different departments. |
| **Chi-Square ($\chi^2$)** | Assessing relationships between categorical variables. | Seeing if there is a relationship between a customer's location (North, South, East, West) and their preferred product category. |

## How It Drives the Final Decision

Once you calculate your test statistic, you use it to make your final statistical decision in one of two ways:

1. **The Critical Value Method:** You compare your test statistic directly to the critical value threshold. If your test statistic is more extreme (further from zero) than the critical value, you reject the null hypothesis.
2. **The P-Value Method:** You convert the test statistic into a p-value using a probability distribution. If the resulting p-value is smaller than your significance level ($\alpha$), you reject the null hypothesis.
