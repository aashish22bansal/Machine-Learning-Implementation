# Alternative Hypothesis

The Alternative Hypothesis (denoted as $H_a$ or $H_1$) is the claim you are actively trying to prove. If the null hypothesis represents the mundane status quo (no effect, no difference), the alternative hypothesis represents discovery—a real shift, relationship, or impact in your data.

When your statistical test yields a very low **p-value**, you reject the null hypothesis and accept the alternative hypothesis as the most plausible explanation for your results.

---

### Direction Matters: One-Tailed vs. Two-Tailed

When formulating an alternative hypothesis, you must decide exactly how specific your claim is. This decision dictates where the "rejection regions" (the threshold for statistical significance) live on your probability curve.

| Test Type | The Claim | Example ($H_a$) | When to Use It |
| --- | --- | --- | --- |
| **Two-Tailed** (Non-directional) | The new variable causes *any* change, whether positive or negative. | "The new marketing strategy's conversion rate is *different* from the old strategy." | You want to catch any effect, even if the new strategy actually performs worse. This is the safest, standard default. |
| **One-Tailed** (Directional) | The new variable moves the metric in one specific, defined direction. | "The new marketing strategy's conversion rate is *greater* than the old strategy." | You only care about an improvement. It requires a strong theoretical justification before running the test. |

> **Key insight:** A one-tailed test makes it mathematically easier to achieve statistical significance in your chosen direction, but it completely blinds you to effects in the opposite direction. If your new strategy is catastrophically worse, a one-tailed test looking only for improvements will treat it the same as "no effect."
