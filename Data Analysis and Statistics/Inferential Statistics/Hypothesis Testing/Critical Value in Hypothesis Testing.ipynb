# Critical Value in Hypothesis Testing

In hypothesis testing, a **critical value** is the exact line in the sand that separates "normal" results from "statistically significant" results.

When you run a test (like a Z-test or T-test), you calculate a **test statistic** from your data. The critical value is the threshold you compare that statistic against to make your final decision.

* If your test statistic is **beyond** the critical value (in the "rejection region"), you reject the null hypothesis ($H_0$).
* If your test statistic is **inside** the critical boundaries, you fail to reject the null hypothesis.

---

### How is it determined?

The critical value is entirely determined by your **Significance Level ($\alpha$)** and whether your test is one-tailed or two-tailed.

If you set your significance level at $\alpha = 0.05$ (meaning you accept a 5% chance of a false positive), the critical value is the exact number on the x-axis of your probability distribution where that 5% area begins.

Here is how critical values change for a standard Normal Distribution (Z-test) based on common $\alpha$ levels:

| Significance Level ($\alpha$) | Confidence Level | One-Tailed Critical Value | Two-Tailed Critical Value |
| --- | --- | --- | --- |
| **0.10** | 90% | $\pm 1.28$ | $\pm 1.645$ |
| **0.05** | 95% | $\pm 1.645$ | $\pm 1.96$ |
| **0.01** | 99% | $\pm 2.33$ | $\pm 2.576$ |

> **Notice the difference:** In a two-tailed test for $\alpha = 0.05$, the 5% rejection region is split in half (2.5% on the left, 2.5% on the right). Because the area is pushed further into the tails, the critical value ($\pm 1.96$) is further from the center than it is for a one-tailed test ($1.645$). This makes it harder to achieve statistical significance in a two-tailed test.

### Critical Value vs. P-Value

Both of these concepts are used to make the same final decision, but they approach it from different angles:

* **Critical Value Method:** Compares the raw test scores. (e.g., *My Z-score is 2.1. The critical value is 1.96. 2.1 is bigger, so I reject $H_0$.*)
* **P-Value Method:** Compares the probabilities. (e.g., *My p-value is 0.035. My $\alpha$ threshold is 0.05. 0.035 is smaller, so I reject $H_0$.*)

Modern software almost always outputs the p-value, making it the more popular method today. However, older statistical tables were built entirely around finding critical values.
