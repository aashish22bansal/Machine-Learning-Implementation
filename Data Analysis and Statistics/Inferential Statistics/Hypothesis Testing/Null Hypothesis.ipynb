# Null Hypothesis

The Null Hypothesis (denoted as $H_0$) is the bedrock of statistical skepticism. It is the assumption that **nothing new, interesting, or significant is happening**—that there is no effect, no difference, and no relationship between the variables you are studying.

Think of it as the statistical equivalent of "innocent until proven guilty." Any difference you observe in your data is assumed to be the result of pure random chance until the evidence is overwhelmingly against it.

---

### How It Shapes Experiments

You always define the null hypothesis in direct opposition to the **Alternative Hypothesis** ($H_a$), which is the actual claim or effect you are trying to prove.

For example, if you are evaluating whether a complex ensemble model predicts outcomes better than a simple baseline model, you don't start by assuming the new model is superior. The null hypothesis dictates the starting line: *The ensemble model's accuracy is exactly equal to the baseline model.* Your objective is to find enough data to reject that claim.

Here is how the null hypothesis looks across different domains:

| Scenario | Null Hypothesis ($H_0$) | Alternative Hypothesis ($H_a$) |
| --- | --- | --- |
| **A/B Testing** | The new website layout has no effect on user conversion rates. | The new website layout increases conversion rates. |
| **Machine Learning** | The new algorithm's error rate is identical to the old algorithm. | The new algorithm reduces the error rate. |
| **Medicine** | The experimental treatment performs exactly the same as a placebo. | The experimental treatment cures the disease faster than a placebo. |

### Why We "Fail to Reject"

In data science and statistics, you never say you "proved" the null hypothesis. The only two valid conclusions are to **reject** $H_0$ or **fail to reject** $H_0$.

This phrasing is highly intentional. If your experiment doesn't show a statistically significant result, it doesn't definitively prove that the two things are identical. It simply means your current data didn't provide enough evidence to prove they are different. When that happens, the status quo ($H_0$) remains the accepted reality.
