# Observation

In statistics, an **observation** (often called a data point, a case, or a record) is a single, specific occurrence of data collection. It represents the complete set of measurements or values collected from *one* individual unit within your sample or population.

If a variable is the *what* you are measuring, an observation is the *who* or *which* you are measuring.

### Visualizing Observations: Rows vs. Columns

The easiest way to understand an observation is to picture a standard spreadsheet or dataset.

* **Columns** represent the **Variables** (the traits you are measuring).
* **Rows** represent the **Observations** (the specific individuals you measured).

| Observation ID | Variable 1 (Age) | Variable 2 (Income) | Variable 3 (Purchased?) |
| --- | --- | --- | --- |
| **Observation 1 (Customer A)** | 25 | $45,000 | Yes |
| **Observation 2 (Customer B)** | 42 | $80,000 | No |
| **Observation 3 (Customer C)** | 31 | $62,000 | Yes |

In this table, the entire second row—Customer B, who is 42 years old, makes $80,000, and did not make a purchase—constitutes exactly one observation.

### Key Concepts Tied to Observations

* **Sample Size ($n$):** In statistical formulas, the total number of observations in your dataset is denoted by the lowercase letter $n$. If you survey 1,000 different people, your sample size is $n = 1000$, meaning you have 1,000 observations.
* **Independence:** Many foundational statistical models and hypothesis tests assume that your observations are strictly independent. This means that Observation 1 has absolutely no influence on Observation 2. For instance, flipping a coin yields independent observations; the first flip doesn't affect the second.
* **Dependent Observations:** When observations are *not* independent, you must use specialized statistical methods. The most common example is **time series data** or **repeated measures**, where you observe the exact same subject multiple times. If you measure a patient's blood pressure every Monday for a month, Observation 2 (Week 2) is heavily dependent on Observation 1 (Week 1) because it is the same patient.
