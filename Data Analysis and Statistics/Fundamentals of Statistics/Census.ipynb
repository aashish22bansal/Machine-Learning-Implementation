# Census

In statistics, a **census** is a method of data collection where you gather information from *every single member* of your target population.

If a sample is just a slice of the pie, a census is the entire pie.

When you conduct a perfect census, you no longer need inferential statistics (like p-values, hypothesis tests, or confidence intervals) because you are not making estimates or educated guesses. You are calculating the exact, true **population parameters** (like the true mean, $\mu$, or the exact population size, $N$).

### When is a Census Used?

While rare due to their scale, a census is typically conducted in two scenarios:

1. **Small, Accessible Populations:** If your population is small enough, a census is completely practical. For example, if you want to know the average salary of the 15 employees in your startup, you don't take a sample of 5 people. You ask HR for the data of all 15.
2. **Government Mandates:** National governments periodically conduct a census (often every 10 years) to count every resident. This data is critical for allocating government funding, determining political representation, and understanding national demographics.

### The Advantages of a Census

* **Absolute Accuracy:** Because you measure everyone, there is zero **sampling error**. The results represent the true state of the population.
* **High Resolution:** It provides detailed information about incredibly small subgroups within the population. A sample might tell you the average income of a country, but a census can tell you the average income of a specific three-block neighborhood.

### The Disadvantages: Why We Rely on Samples

If a census provides perfect information, why do data scientists and statisticians spend their entire careers looking at samples?

* **Prohibitive Cost and Time:** Surveying millions of people requires massive amounts of money, manpower, and years to collect, process, and clean the data. By the time a national census is published, the data is often already outdated.
* **Logistical Impossibility:** For infinite populations (like "all future customers") or naturally elusive populations (like wildlife in a rainforest), a true census is physically impossible.
* **Destructive Testing:** In manufacturing, testing the lifespan of a lightbulb requires leaving it on until it burns out. A census would mean destroying the entire inventory.
* **Non-Response Bias:** Even in a government census, it is impossible to reach 100% of the population. Homeless individuals, undocumented immigrants, or people who simply refuse to answer the door create a gap in the data that can skew the "perfect" result.

### Census vs. Sample Comparison

| Feature | Census | Sample |
| --- | --- | --- |
| **Definition** | Data collected from the entire population. | Data collected from a representative subset. |
| **Result** | Yields a **Parameter** (the true value). | Yields a **Statistic** (an estimate). |
| **Time & Cost** | Extremely high. | Relatively low. |
| **Sampling Error** | None. | Always present (measured by margin of error). |
| **Best For** | Small populations, or when absolute precision across all subgroups is legally required. | Large or infinite populations where speed, cost, and practicality are priority. |
