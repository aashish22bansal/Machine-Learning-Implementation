# Parameter

In statistics, a **parameter** is a specific numerical value that describes a characteristic of an entire **population**.

If a population is the complete, comprehensive set of all individuals or items you want to study, the parameter is the absolute, unvarnished truth about that group. If you could somehow measure every single entity in the population with zero error, the number you calculate would be the parameter.

### The Ultimate Goal of Statistics

Parameters represent the "holy grail" of inferential statistics. When a data scientist runs a complex machine learning model or a statistician conducts a hypothesis test, their ultimate goal is to understand the true parameter.

However, because populations are usually massive (or infinite), **parameters are almost always unknown and practically unknowable**. You cannot survey every single voter, test every single manufactured part, or predict the behavior of every future user. Therefore, we use a sample statistic to estimate the hidden population parameter.

### Notation: The Greek Alphabet

To ensure everyone knows whether they are looking at a true population value or just an estimated sample value, statisticians use Greek letters for parameters and standard Latin letters for statistics.

| Metric | Population Parameter (The True Value) | Sample Statistic (The Estimate) |
| --- | --- | --- |
| **Mean (Average)** | $\mu$ (Mu) | $\bar{x}$ (x-bar) |
| **Standard Deviation** | $\sigma$ (Sigma) | $s$ |
| **Variance** | $\sigma^2$ | $s^2$ |
| **Proportion** | $p$ (or $\pi$) | $\hat{p}$ (p-hat) |
| **Correlation** | $\rho$ (Rho) | $r$ |
| **Size** | $N$ | $n$ |

### Key Characteristics of a Parameter

1. **It is a Fixed Constant:** In standard (frequentist) statistics, the population parameter does not change. For example, at this exact millisecond, there is a true, fixed average height for all adult women on Earth ($\mu$). We don't know what that exact number is, but the number itself exists and is constant.
2. **It Has No Error:** Unlike a sample statistic, which fluctuates depending on who you happen to survey (sampling error), a parameter has zero sampling error because it includes absolutely everyone.
3. **It Requires a Census:** The only way to perfectly calculate a parameter rather than estimating it is to conduct a true, 100% complete census of the population.

### The Bridge Between Statistic and Parameter

Because we cannot see the parameter, we build a mathematical bridge to it using **Confidence Intervals**.

When a researcher states, "We are 95% confident that the true average conversion rate is between 4.2% and 4.8%," they are explicitly acknowledging that the parameter ($p$) is unknown. They are using their sample statistic ($\hat{p}$) to draw a boundary where they are mathematically confident the true, fixed parameter actually lives.
