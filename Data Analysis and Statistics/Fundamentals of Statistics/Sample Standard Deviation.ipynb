# Sample Standard Deviation

In statistics, **sample standard deviation** is the ultimate, practical measurement of how spread out your data is.

If you recall from the previous concept, sample variance ($s^2$) is mathematically essential but practically useless because it is expressed in "squared units" (e.g., squared dollars or squared centimeters).

The sample standard deviation completely fixes this problem. It is simply the square root of the sample variance, which brings the measurement perfectly back into the original units of your data. It tells you, on average, exactly how far each data point deviates from the sample mean.

To distinguish it from the population standard deviation ($\sigma$), statisticians denote the sample standard deviation using the lowercase letter **$s$**.

### The Mathematical Formula

Because it is just the square root of the sample variance, the formula incorporates the exact same logic, including Bessel's Correction ($n - 1$) in the denominator to ensure it remains an unbiased estimator of the true population spread.

$$s = \sqrt{\frac{\sum (x - \bar{x})^2}{n - 1}}$$

* **$s$** = Sample standard deviation
* **$\sum$** = The sum of...
* **$x$** = Each individual observation
* **$\bar{x}$** = The sample mean
* **$n - 1$** = The sample size minus one

### Why It Matters: The Empirical Rule

The true superpower of the standard deviation unlocks when your data follows a normal distribution (a bell curve). Standard deviation acts as a universal yardstick.

If you know the mean ($\bar{x}$) and the standard deviation ($s$), you can instantly predict where almost all of your data will fall using the **Empirical Rule** (also known as the 68-95-99.7 rule):

image.png

*The Empirical Rule (68-95-99.7). Source: PeterHermesFurian / Getty Images*

* **~68%** of all data points fall within **1 standard deviation** ($\pm 1s$) of the mean.
* **~95%** of all data points fall within **2 standard deviations** ($\pm 2s$) of the mean.
* **~99.7%** of all data points fall within **3 standard deviations** ($\pm 3s$) of the mean.

**Example Scenario:**
Imagine you are analyzing the lifespan of a specific battery model in your sample.

* **Mean ($\bar{x}$):** 50 hours
* **Standard Deviation ($s$):** 2 hours

Because you have the standard deviation in real units (hours), you can immediately tell your engineering team that 95% of these batteries will last between **46 hours and 54 hours** (the mean $\pm$ 2 standard deviations). If a battery dies at 43 hours, it is more than 3 standard deviations away—a rare outlier.

### Sample vs. Population Standard Deviation

Just like with variance, it is critical to know whether you have data for the entire population or just a sample.

| Feature | Sample Standard Deviation | Population Standard Deviation |
| --- | --- | --- |
| **Notation** | $s$ | $\sigma$ (Sigma) |
| **Denominator** | $n - 1$ | $N$ |
| **Use Case** | Estimating the spread of a larger group based on subset data. | Calculating the exact spread when you have data for every single member. |
