# Statistic

In the field of statistics, a **statistic** is a specific numerical value calculated from a **sample** of data.

Its primary job is to summarize the data within that sample and, more importantly, to serve as an estimate for the corresponding true value in the broader population (which is called a parameter).

If you survey $1,000$ voters out of a country of $50$ million and find that $54\%$ support a specific policy, that $54\%$ is a statistic. It is a mathematical fact about your specific sample of $1,000$ people, and an educated guess about the $50$ million.

### Statistic vs. Parameter

Understanding the difference between a statistic and a parameter is one of the most fundamental concepts in data science. The easiest way to remember the difference is through alliteration:

* **S**tatistics come from **S**amples.
* **P**arameters come from **P**opulations.

To keep the math clear, statisticians use different alphabets to denote them. Statistics are typically represented by standard Latin letters, while parameters use Greek letters.

| Metric | Sample Statistic (The Estimate) | Population Parameter (The True Value) |
| --- | --- | --- |
| **Mean (Average)** | $\bar{x}$ (x-bar) | $\mu$ (Mu) |
| **Standard Deviation** | $s$ | $\sigma$ (Sigma) |
| **Variance** | $s^2$ | $\sigma^2$ |
| **Proportion / Percentage** | $\hat{p}$ (p-hat) | $p$ (or $\pi$) |
| **Size / Count** | $n$ | $N$ |

### The Two Branches of Statistics

The word "statistic" also broadly refers to the two main branches of the discipline itself, which dictate how you use these calculated numbers:

#### 1. Descriptive Statistics

This branch uses statistics solely to describe, summarize, and organize the data you have actually collected. It does not try to guess anything beyond the data in front of you.

* *Example:* Calculating the average age, the median income, or the standard deviation of test scores for a specific classroom of 30 students. You are just stating facts about those 30 students.

#### 2. Inferential Statistics

This branch takes the descriptive statistic calculated from a sample and uses it to make predictions, inferences, or generalizations about the larger population. Because you are guessing, inferential statistics always involve uncertainty.

* *Example:* Taking the average test score of those 30 students ($\bar{x}$) and using it to estimate the average test score of all $10,000$ students in the district ($\mu$), while calculating a margin of error (Confidence Interval) and testing for significance (Hypothesis Testing).

### The Role of Sampling Error

Because a statistic is calculated from a subset of the population, it will almost never perfectly match the true population parameter. If the true average height of a population is $170\text{ cm}$, your sample statistic might be $170.2\text{ cm}$ or $169.5\text{ cm}$.

This natural gap between the sample statistic and the true population parameter is called **sampling error**. It is not a mistake; it is a mathematical reality. The entire toolkit of inferential statistics exists to measure that exact error and ensure your sample statistic is a reliable, trustworthy estimate.
