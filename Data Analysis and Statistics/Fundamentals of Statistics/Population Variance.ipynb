# Population Variance

In statistics, **population variance** is a crucial measure of dispersion. While the mean tells you where the center of your data is, the variance tells you how widely spread out the data points are around that center.

Specifically, population variance calculates the **average squared distance of every single data point from the true population mean**.

Because it describes an entire population rather than a sample, it is a fixed **parameter**. To distinguish it from sample variance, statisticians denote population variance using the Greek letter sigma squared (**$\sigma^2$**).

### The Mathematical Formula

The formula to calculate the population variance involves looking at every single observation in the population:

$$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$

* **$\sigma^2$** = Population variance
* **$\sum$** = The sum of...
* **$x$** = Each individual observation in the population
* **$\mu$** = The population mean
* **$N$** = The total population size

To calculate it step-by-step: you find the mean ($\mu$), subtract the mean from every single data point to find the "deviation" ($x - \mu$), square each of those deviations, add all those squared numbers together, and finally divide by the total number of items ($N$).

### Why Do We Square the Distances?

Looking at the formula, you might wonder why we don't just take the average of the regular, un-squared distances from the mean. There are two mathematical reasons for this:

1. **Preventing Cancellation:** By definition, some data points will be above the mean (positive distance) and some will be below the mean (negative distance). If you simply add them all up without squaring them, the positive and negative numbers perfectly cancel each other out, always equaling zero. Squaring them makes every distance a positive number.
2. **Penalizing Outliers:** Squaring the distances gives exponentially more weight to extreme values. A data point that is 4 units away from the mean adds 16 to the total variance, while a point only 1 unit away adds just 1. This ensures that variance is highly sensitive to outliers.

### The "Squared Units" Problem

While variance is mathematically foundational and used constantly in probability equations and machine learning algorithms, it is highly unintuitive for practical, everyday interpretation.

Because you squared the distances, **the variance is always expressed in squared units.**

* If you are measuring human height in centimeters (cm), the variance is in **cm²**.
* If you are measuring employee salaries in dollars, the variance is in **squared dollars**.

To make the spread understandable in real-world terms, statisticians take the square root of the variance to calculate the **Population Standard Deviation ($\sigma$)**. This brings the measurement perfectly back into the original units.

### Population Variance ($\sigma^2$) vs. Sample Variance ($s^2$)

If you are calculating the variance of an entire population, you divide by the exact total number of items ($N$).

However, if you are only working with a sample, calculating the variance using $n$ (the sample size) in the denominator will consistently underestimate the true spread of the population. To fix this bias, the formula for **Sample Variance ($s^2$)** uses what is known as Bessel's correction, dividing by **$n - 1$** instead of $N$.
