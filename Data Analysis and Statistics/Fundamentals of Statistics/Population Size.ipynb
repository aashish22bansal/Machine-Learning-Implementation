# Population Size

In statistics, **population size** is the total number of individuals, items, or observations that make up the entire group you are studying. It represents the absolute maximum number of data points that could possibly be collected if you were to conduct a perfect census.

To distinguish it from the size of a smaller sample, statisticians universally denote the total population size using a capital letter **$N$**, while the sample size is denoted by a lowercase **$n$**.

### Finite vs. Infinite Populations

The exact size of a population fundamentally changes how you approach data collection and statistical formulas. Populations fall into two mathematical categories:

* **Finite Populations:** The population size ($N$) is a distinct, countable number, even if it is very large.
* *Example:* If you are studying the voting habits of registered voters in a specific city, and there are exactly 124,500 registered voters, then $N = 124500$.


* **Infinite (or Hypothetical) Populations:** The population size is theoretically limitless, uncountable, or represents a continuous process, making $N$ effectively infinite ($\infty$).
* *Example:* The population of "all possible outcomes of rolling a six-sided die" or "all future customers who will ever visit a website."



### How Population Size Affects the Math

A common misconception is that you need a larger sample size ($n$) to accurately measure a larger population size ($N$). In reality, once a population reaches a certain size, the mathematical requirements for your sample size barely change.

Whether a country has 10 million people or 300 million people, a well-randomized sample of about 1,000 people will yield a nearly identical margin of error (roughly $\pm 3\%$) for both populations. The absolute size of the sample matters far more than the fraction of the population it represents.

However, population size does matter mathematically in one specific scenario: **The Finite Population Correction (FPC).**

#### The Finite Population Correction Factor

Standard statistical formulas (like calculating the standard error) assume that your population is infinite. If your population is finite and your sample makes up a significantly large chunk of that total population (usually defined as **more than 5%**, or $n > 0.05N$), the standard formulas become overly pessimistic and overestimate your error.

When you sample a large fraction of a finite population, you have less uncertainty. To adjust for this, statisticians multiply the standard error by the FPC factor:

$$\text{FPC} = \sqrt{\frac{N - n}{N - 1}}$$

* **$N$** = Total Population Size
* **$n$** = Sample Size

If $N$ is massive (like 300 million) and $n$ is small (like 1,000), this formula essentially equals 1, meaning it has no effect on the standard error. But if $N$ is 1,000 and you sample 500 of them, the FPC shrinks your calculated error, mathematically acknowledging that because you measured half of the entire population, your estimate is highly precise.
