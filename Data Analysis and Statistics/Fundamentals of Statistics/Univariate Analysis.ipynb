# Univariate Analysis

In statistics, **univariate analysis** is the simplest form of analyzing data. As the prefix "uni" suggests, it involves the examination of just **one single variable** at a time.

Because you are only looking at one variable, univariate analysis does not deal with causes, effects, or relationships. You cannot use it to see if fertilizer increases plant growth, or if age affects income. Its sole purpose is to describe the data, summarize it, and find patterns within that specific single slice of information.

Think of it as introducing yourself to a dataset one column at a time before you start trying to connect the dots.

---

### What Are We Looking For?

When performing univariate analysis on a numeric variable, statisticians are generally trying to answer three core questions:

1. **Central Tendency:** Where is the middle of the data? (Measured by the Mean, Median, and Mode).
2. **Dispersion (Spread):** How spread out are the data points around that middle? (Measured by Range, Variance, Standard Deviation, and Interquartile Range).
3. **Shape & Distribution:** Is the data perfectly symmetrical, or is it skewed by extreme outliers? (Measured by Skewness, Kurtosis, and visualizing the probability distribution).

### Methods Based on Variable Type

The techniques you use for univariate analysis depend entirely on whether your single variable is categorical (labels/categories) or quantitative (numbers).

| Variable Type | Common Statistical Summaries | Best Visualizations | Example Question |
| --- | --- | --- | --- |
| **Categorical** (e.g., Eye Color, Product Categories) | Frequencies (counts), Proportions, Percentages, Mode (most common category). | Bar Charts, Pie Charts, Frequency Tables. | "What percentage of our customers are located in the North region?" |
| **Quantitative** (e.g., Age, Revenue, Temperature) | Mean, Median, Standard Deviation ($s$), Variance ($s^2$), Minimum, Maximum, Quartiles. | Histograms, Box Plots, Density Curves. | "What is the average transaction value, and how much does it vary?" |

### Why It Is the Crucial First Step

Univariate analysis is the bedrock of **Exploratory Data Analysis (EDA)**. Before you ever attempt complex machine learning algorithms or bivariate hypothesis testing (comparing two variables), you must run a univariate analysis on every single variable in your dataset.

Here is why:

* **Catching Errors:** If you run a summary on an "Age" variable and see a maximum value of 250, you instantly know there is a data entry error.
* **Finding Outliers:** A box plot will immediately show you if one massive data point is going to skew your entire predictive model.
* **Checking Assumptions:** Many statistical tests (like the Z-test or T-test) assume your data follows a normal distribution. A univariate histogram is the quickest way to check if your data actually looks like a bell curve.

Once you fully understand how each variable behaves on its own through univariate analysis, you can confidently move on to **Bivariate Analysis** (comparing two variables) and **Multivariate Analysis** (comparing three or more).
