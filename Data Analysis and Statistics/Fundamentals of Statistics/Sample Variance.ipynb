# Sample Variance

In statistics, **sample variance** is the mathematical measurement of how spread out your sample data is around the sample mean.

Just like population variance, it measures the average squared distance of data points from the center. However, because it is calculated from a subset of data (a sample) rather than the entire population, it serves as an **estimator** for the true, unknown population variance.

To distinguish it from population variance ($\sigma^2$), statisticians universally denote sample variance using the lowercase letter **$s^2$**.

### The Mathematical Formula

The formula for sample variance looks nearly identical to population variance, with one massive, mathematically crucial difference in the denominator:

$$s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$$

* **$s^2$** = Sample variance
* **$\sum$** = The sum of...
* **$x$** = Each individual observation in your sample
* **$\bar{x}$** = The sample mean
* **$n$** = The sample size (total number of observations)

### The Mystery of $n - 1$ (Bessel's Correction)

The most common question people ask when looking at this formula is: *"Why do we divide by $n - 1$ instead of just $n$?"*

If you are trying to find an average, shouldn't you divide by the total number of items?

If you divide by $n$, you are calculating the exact variance of *your specific sample*. However, data scientists almost never care about the sample itself; they care about what the sample tells them about the **population**.

1. **The Built-In Bias:** Your sample data is always going to cluster slightly closer to its *own* sample mean ($\bar{x}$) than it does to the true, unknown population mean ($\mu$).
2. **The Underestimation:** Because the data is artificially tightly packed around $\bar{x}$, dividing by $n$ will consistently **underestimate** the true spread of the population.
3. **The Correction:** To fix this bias, statisticians use what is known as **Bessel's Correction**. By shrinking the denominator from $n$ to $n - 1$, you force the final mathematical result to be a little bit larger.

This tiny artificial inflation perfectly corrects the bias, making $s^2$ an **unbiased estimator** of the true population variance.

*(Note: As your sample size $n$ gets larger—say, moving from 10 to 1,000—the difference between dividing by $n$ and $n-1$ shrinks to almost zero. Bessel's correction is most vital for small samples.)*

### Sample vs. Population Variance

| Feature | Sample Variance | Population Variance |
| --- | --- | --- |
| **Notation** | $s^2$ | $\sigma^2$ |
| **Center Point Used** | Sample Mean ($\bar{x}$) | Population Mean ($\mu$) |
| **Denominator** | $n - 1$ (Bessel's Correction) | $N$ (Total population size) |
| **Purpose** | To *estimate* the true spread of the population. | To state the *exact* true spread of the population. |

### The "Squared Units" Problem

Just like population variance, sample variance is expressed in squared units because the formula squares the distances to prevent positive and negative numbers from canceling out.

If you are measuring a sample of employee salaries in dollars, your sample variance will be in **squared dollars**—a metric that is mathematically useful but practically impossible to interpret.
