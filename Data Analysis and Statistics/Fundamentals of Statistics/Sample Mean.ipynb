# Sample Mean

In statistics, the **sample mean** is the mathematical average of a specific subset of data (a sample) drawn from a larger population.

Because it is calculated from a sample rather than the entire group, it is a **statistic**, not a parameter. It serves as our primary "point estimate"—our mathematically best guess—for the true, unknown average of the entire population.

To distinguish it from the population mean ($\mu$), statisticians universally denote the sample mean using the symbol **$\bar{x}$** (pronounced "x-bar").

### The Mathematical Formula

The formula for the sample mean is nearly identical in structure to the population mean, but it uses sample-specific notation:

$$\bar{x} = \frac{\sum x}{n}$$

* **$\bar{x}$** = The sample mean
* **$\sum x$** = The sum of all individual observations in your specific sample
* **$n$** = The sample size (the total number of observations you collected)

To calculate it, you simply add up the values of every data point in your sample and divide by the total number of data points you collected.

### The Role of the Sample Mean in Inference

In the real world, calculating the exact average of an entire population ($\mu$) is usually impossible. You cannot measure the exact height of every human or the exact lifespan of every lightbulb produced in a factory. The sample mean ($\bar{x}$) exists to bridge that gap.

However, because you are only measuring a fraction of the population, **$\bar{x}$ will almost never perfectly equal $\mu$.** If the true average height of a population is $170\text{ cm}$, your specific sample of 100 people might average $170.4\text{ cm}$. If you take a second sample of 100 different people, their average might be $169.2\text{ cm}$.

This natural wobble between different sample means is called **sampling error**.

### The Central Limit Theorem (CLT)

The sample mean is the star player of the most important theorem in all of statistics: **The Central Limit Theorem**.

The CLT states that if you take many appropriately sized samples (usually $n \ge 30$) from *any* population and calculate the sample mean ($\bar{x}$) for each one, those sample means will form a perfect normal distribution (a bell curve). Most importantly, the exact center of that bell curve will be the true population mean ($\mu$).

This mathematical phenomenon allows data scientists to take just *one* sample mean, calculate its standard error, and build a Confidence Interval to state exactly how close their estimate is likely to be to the true population average.

### Sample Mean vs. Population Mean

| Feature | Sample Mean | Population Mean |
| --- | --- | --- |
| **Notation** | $\bar{x}$ (x-bar) | $\mu$ (Mu) |
| **Classification** | Statistic (An estimate) | Parameter (The exact truth) |
| **Denominator Used** | $n$ (Sample size) | $N$ (Total population size) |
| **Error** | Contains sampling error. | Contains zero sampling error. |
| **Availability** | Easily calculated from collected data. | Usually unknown and practically unknowable. |
