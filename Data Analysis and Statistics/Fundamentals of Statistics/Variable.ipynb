# Variable

In statistics, a **variable** is any characteristic, number, or quantity that can be measured or counted.

It is called a "variable" simply because its value can vary between different individuals in a population, or it can change over time. If you are conducting a study on a group of people, characteristics like their age, eye color, income, and resting heart rate are all variables.

### Types of Variables

Variables dictate exactly what kind of statistical analysis you can perform. They are broadly divided into two main categories, which are then broken down further:

| Category | Type | Definition | Examples |
| --- | --- | --- | --- |
| **Categorical (Qualitative)** | **Nominal** | Variables that represent discrete categories with no inherent mathematical order or ranking. | Blood type, hair color, country of birth, dog breed. |
| **Categorical (Qualitative)** | **Ordinal** | Variables that represent categories with a clear, logical progression or rank, but the exact distance between ranks is unknown. | Customer satisfaction (Poor, Fair, Good), education level, military ranks. |
| **Numeric (Quantitative)** | **Discrete** | Variables that can only take on specific, countable, distinct values (usually whole numbers). | Number of children in a family, number of cars passing a tollbooth, shoe size. |
| **Numeric (Quantitative)** | **Continuous** | Variables that can take on any possible mathematical value within a range, including infinite fractions or decimals. | Exact height, weight, time it takes to run a marathon, temperature. |

### The Role of Variables in Experiments

When you move from simply describing data to actually testing hypotheses or building predictive models, variables take on specific roles based on how they interact with each other.

In any experiment or machine learning model, you are usually looking at the relationship between two main types of variables:

* **Independent Variable (Predictor or Feature):** This is the variable that you (the researcher) manipulate, or the input data you are using to make a prediction. It is the presumed *cause*. It is often plotted on the x-axis of a graph.
* **Dependent Variable (Response or Target):** This is the variable you are trying to measure, predict, or explain. It is the presumed *effect*, and its value depends entirely on changes in the independent variable. It is often plotted on the y-axis.

**Example Scenario:**
If you are testing how different amounts of fertilizer affect the growth rate of a plant:

* The **Independent Variable** is the exact amount of fertilizer applied (you control this).
* The **Dependent Variable** is the height the plant reaches (you measure this to see the effect).
* **Confounding Variables** are outside factors (like sunlight or soil quality) that you must control, otherwise, they might secretly influence the dependent variable and ruin your statistical results.
