# Sample

In statistics, a **sample** is a smaller, manageable subset of individuals or data points extracted from a larger population.

If measuring the entire population is impossible, too expensive, or too time-consuming (making a census impractical), statisticians rely on a sample to act as a stand-in for the whole. The entire field of **inferential statistics**—including hypothesis testing, p-values, and confidence intervals—is dedicated to figuring out how accurately a sample reflects the true population.

### Sample Statistics vs. Population Parameters

When you calculate a metric using only your sample data, that resulting number is called a **statistic**. A statistic is always an estimate or a "best guess" of the true population parameter.

To keep the math clear, statisticians use different symbols to distinguish between the two:

| Metric | Sample Statistic (The Estimate) | Population Parameter (The True Value) |
| --- | --- | --- |
| **Mean (Average)** | $\bar{x}$ (x-bar) | $\mu$ (Mu) |
| **Standard Deviation** | $s$ | $\sigma$ (Sigma) |
| **Variance** | $s^2$ | $\sigma^2$ |
| **Size/Count** | $n$ | $N$ |

For example, if you want to know the average accuracy of a machine learning model across millions of potential inputs, you might test it on a sample of $1,000$ inputs ($n = 1,000$). The average accuracy of that specific test is your sample mean ($\bar{x}$). You then use statistical methods to infer how close $\bar{x}$ is likely to be to the true, universal accuracy ($\mu$).

### The Golden Rule: Representativeness

The sheer size of a sample matters far less than its **quality**. A sample of $10,000$ data points is completely useless if it is heavily biased.

To make valid inferences, a sample must be **representative**, meaning its characteristics closely mirror the characteristics of the overall population. If the population is $50\%$ female and $50\%$ male, a representative sample should roughly maintain that same ratio.

### Common Types of Sampling

How you choose your sample dictates how much you can trust your resulting statistics.

* **Simple Random Sampling:** Every single member of the population has an equal, mathematically fair chance of being selected (like pulling names from a hat). This is the gold standard for avoiding bias.
* **Stratified Sampling:** The population is divided into distinct subgroups (strata), and random samples are taken proportionally from each group. This ensures minority groups are adequately represented.
* **Convenience Sampling:** Gathering data from whatever sources are easiest to reach (e.g., polling your friends, or scraping the first 100 pages of a website). This almost always introduces severe bias and invalidates most formal statistical tests.

### Sampling Error

Because a sample is only a fraction of the population, it will never be perfectly accurate. This natural, expected discrepancy between the sample statistic ($\bar{x}$) and the true population parameter ($\mu$) is called **sampling error**.

Sampling error is not a mistake or a flaw in the methodology; it is just a mathematical reality of not surveying everyone. This is exactly why we use **Confidence Intervals**—to mathematically quantify the likely size of that sampling error and draw a boundary around our estimates.
