# Size of the Dataset

When discussing frequency distributions, the absolute count of an event doesn't mean much until you know the **total size of the dataset**. Finding 50 defective parts in a factory is a disaster if you only inspected 100 parts, but it is an incredible success if you inspected 10 million.

In statistics, the size of the dataset is formally denoted by two specific letters, depending on what the data represents:

| Symbol | Meaning | Example |
| --- | --- | --- |
| **$N$** (Capital N) | **Population Size:** The total number of *all possible* observations in the entire group you are studying. | Every single voter in a country ($N$ = 150,000,000). |
| **$n$** (Lowercase n) | **Sample Size:** The number of observations in a smaller subset drawn from that population. | A telephone poll of voters ($n$ = 1,000). |

## Why Dataset Size Matters for Frequency

The size of your dataset ($N$ or $n$) is the mathematical foundation for converting raw counts into useful insights.

1. **Calculating Relative Frequency:** To understand the proportion of an event, you must divide the absolute frequency (the raw count, $f$) by the total dataset size ($n$).
* *Formula:* Relative Frequency = $f / n$


2. **Determining Statistical Power:** In inferential statistics, a larger $n$ means your frequency distribution is more likely to be an accurate reflection of the true population. Small samples are highly susceptible to random chance and outliers.

---

## The Law of Large Numbers

The most important relationship between dataset size and frequency is the **Law of Large Numbers**. This mathematical theorem states that as your sample size ($n$) grows larger, the *relative frequency* of an outcome will get closer and closer to the *true population frequency* (the expected probability).

If you flip a coin 10 times ($n=10$), you might get 8 heads (a relative frequency of 80%). But if you flip it 10,000 times ($n=10,000$), your relative frequency will tightly converge around the true probability of 50%.

{/* Reason: An interactive simulation is the best way to intuitively grasp the Law of Large Numbers—watching the chaotic frequency of small samples stabilize as 'n' grows. */}

This Colab engine simulates flipping a coin. It runs three independent simulations simultaneously so you can see how they all start out chaotic, but eventually converge on the exact same true probability (0.5) as the dataset size grows.

````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def simulate_lln(max_flips):
    """
    Simulates coin flips to demonstrate the Law of Large Numbers.
    Runs 3 independent simulations to show convergence.
    """
    num_simulations = 3

    plt.figure(figsize=(12, 6))

    # Draw the true probability target line (50% for a fair coin)
    plt.axhline(y=0.5, color='black', linestyle='--', linewidth=2, zorder=10, label='True Probability (0.5)')

    colors = ['#e74c3c', '#3498db', '#2ecc71']

    for i in range(num_simulations):
        # Generate random flips (1 = Heads, 0 = Tails)
        flips = np.random.randint(0, 2, size=max_flips)

        # Calculate the cumulative number of heads at each flip
        cumulative_heads = np.cumsum(flips)

        # Create an array representing the sample size 'n' at each step [1, 2, 3, ... max_flips]
        n_trials = np.arange(1, max_flips + 1)

        # Calculate the running relative frequency (Cumulative Heads / n)
        running_relative_frequency = cumulative_heads / n_trials

        # Plot the running frequency
        plt.plot(n_trials, running_relative_frequency, color=colors[i], alpha=0.8,
                 linewidth=1.5, label=f'Simulation {i+1}')

    # Formatting the chart
    plt.title(f"Law of Large Numbers: Convergence over {max_flips:,} Flips", fontsize=15, pad=15)
    plt.xlabel("Sample Size (n) - Number of Flips", fontsize=12)
    plt.ylabel("Relative Frequency of 'Heads'", fontsize=12)

    # Dynamically adjust the Y-axis to zoom in as the sample size gets larger
    if max_flips <= 100:
        plt.ylim(0.0, 1.0)
    elif max_flips <= 1000:
        plt.ylim(0.35, 0.65)
    else:
        plt.ylim(0.45, 0.55)

    # Use a logarithmic scale for the X-axis.
    # This stretches out the low 'n' values so you can clearly see the early chaos.
    plt.xscale('log')

    plt.legend(loc='upper right')
    plt.grid(True, which="both", ls="--", alpha=0.5)

    plt.tight_layout()
    plt.show()

print("Adjust the slider to increase the sample size (n) from 10 to 100,000.")
print("Notice how the relative frequency stabilizes as the dataset grows.")

# Create an interactive slider for the exponent (1 to 5)
# This will control the 'power' for the 10^power number of flips.
exponent_slider = widgets.IntSlider(
    value=2,  # Corresponds to 10^2 = 100
    min=1,    # Exponent for 10^1 = 10
    max=5,    # Exponent for 10^5 = 100,000
    step=1,
    description='Max Flips (10^X):', # Updated description to clarify it's an exponent
    continuous_update=False,
    layout=widgets.Layout(width='500px')
)

# Bind the slider to the plot function, calculating max_flips from the exponent
widgets.interact(lambda exponent: simulate_lln(10**exponent), exponent=exponent_slider);
````

### What to Look For in the Output

* **Early Chaos (Low $n$):** Look at the far left side of the chart where $n$ is under 100. The lines swing wildly between 0.2 and 0.8. A dataset this small gives an unreliable measure of frequency.
* **The Squeeze (Medium $n$):** As $n$ approaches 1,000, the extreme swings disappear, but the lines still wander slightly above or below the 0.5 mark.
* **Absolute Convergence (High $n$):** As you push the slider to 10,000 or 100,000, the three independent simulation lines will squeeze tightly against the black dotted 0.5 line. The dataset is now large enough that the relative frequency perfectly mirrors the true mathematical probability.
