# Percentage

Building directly on the concepts of "Dataset Size" and "Count," **percentage** is the mathematical bridge that allows you to compare different datasets fairly.

If a hospital treats 50 patients successfully, is that a good outcome? You cannot know until you know the dataset size ($n$). If they treated 50 out of 50 patients, it is excellent. If they treated 50 out of 5,000, it is terrible.

A percentage mathematically standardizes raw counts by expressing them as a fraction of 100.

$$\text{Percentage} = \left( \frac{\text{Part}}{\text{Whole}} \right) \times 100$$

## Types of Percentages in Data

In descriptive statistics, you will encounter percentages used in two entirely distinct ways. Mixing these up is a common source of data misinterpretation.

### 1. Relative Frequency (Proportion of a Whole)

This is the most common use. It describes how much of a single dataset belongs to a specific category.

* *Example:* "60% of the respondents prefer tea over coffee."
* *Rule:* If you are measuring mutually exclusive categories within a single dataset, the relative frequencies must always sum exactly to 100%.

{/* Reason: An interactive chart suite allows users to switch between pie, donut, and bar charts to see how proportions of a whole (relative frequency) are best visualized, and how they must always sum to 100%. */}

### 2. Percentage Change (Difference Over Time)

This measures how a single variable has grown or shrunk compared to a previous state. Unlike relative frequency, percentage change is not capped at 100%—a startup's revenue can grow by 400%.

$$\text{Percentage Change} = \left( \frac{\text{New Value} - \text{Old Value}}{\text{Old Value}} \right) \times 100$$

* *Example:* If a store sold 10 shirts yesterday and 15 shirts today, the percentage change is $\left( \frac{15 - 10}{10} \right) \times 100 = 50\%$ increase.

## Common Statistical Pitfalls with Percentages

Because percentages look like simple, standardized numbers, they are very easy to manipulate or misinterpret:

* **The Small Sample Illusion:** "100% of customers surveyed were satisfied!" sounds amazing, but if $n = 2$, the statistic is functionally useless. Always demand the raw count before trusting a percentage.
* **Averaging Percentages:** You cannot mathematically average two percentages unless their underlying raw counts (dataset sizes) are identical. If Group A has an 80% pass rate (out of 10 people) and Group B has a 40% pass rate (out of 1,000 people), the overall average is *not* 60%. It is heavily weighted toward Group B's performance.
* **Percentage vs. Percentage Point:** If an interest rate goes from 10% to 12%, it did not increase by 2%. It increased by **2 percentage points**. The actual *percentage change* of the rate itself is a 20% increase (2 is 20% of 10).

This Python script can visualize how relative frequencies (parts of a whole) must always sum to exactly 100%.

This Colab engine allows you to adjust the proportions of two groups, automatically calculates the third to maintain the 100% boundary, and lets you toggle between different chart types to see how each visualizes the same data differently.

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def update_proportions(group_a, group_b, chart_type):
    """
    Calculates the 100% boundary, adjusts group values if they exceed it,
    and draws the selected chart type.
    """
    # Enforce the 100% rule: Group A + Group B cannot exceed 100
    if group_a + group_b > 100:
        group_b = 100 - group_a

    # Group C auto-calculates to fill the remainder
    group_c = 100 - group_a - group_b

    # Setup data
    labels = ['Group A', 'Group B', 'Group C']
    sizes = [group_a, group_b, group_c]
    colors = ['#3498db', '#e74c3c', '#2ecc71']

    # Create the plot
    fig, ax = plt.subplots(figsize=(10, 6))

    if chart_type == 'Pie Chart':
        # Standard Pie Chart
        wedges, texts, autotexts = ax.pie(
            sizes, labels=labels, autopct='%1.1f%%', startangle=90,
            colors=colors, wedgeprops={'edgecolor': 'white', 'linewidth': 2},
            textprops={'fontsize': 12, 'fontweight': 'bold'}
        )
        ax.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
        ax.set_title("Relative Frequency: Pie Chart", fontsize=15, pad=20)

    elif chart_type == 'Donut Chart':
        # Donut chart is just a pie chart with a white circle in the center
        wedges, texts, autotexts = ax.pie(
            sizes, labels=labels, autopct='%1.1f%%', startangle=90, pctdistance=0.85,
            colors=colors, wedgeprops={'edgecolor': 'white', 'linewidth': 2},
            textprops={'fontsize': 12, 'fontweight': 'bold'}
        )
        # Draw the center circle
        centre_circle = plt.Circle((0, 0), 0.65, fc='white')
        ax.add_artist(centre_circle)
        ax.axis('equal')
        ax.set_title("Relative Frequency: Donut Chart", fontsize=15, pad=20)

    elif chart_type == 'Stacked Bar Chart':
        # 100% Stacked Bar Chart
        ax.barh('Total Population', group_a, color=colors[0], edgecolor='white', label='Group A')
        ax.barh('Total Population', group_b, left=group_a, color=colors[1], edgecolor='white', label='Group B')
        ax.barh('Total Population', group_c, left=group_a + group_b, color=colors[2], edgecolor='white', label='Group C')

        # Add percentage text inside the bars
        if group_a > 3:
            ax.text(group_a/2, 0, f"{group_a}%", ha='center', va='center', color='white', fontweight='bold', fontsize=12)
        if group_b > 3:
            ax.text(group_a + group_b/2, 0, f"{group_b}%", ha='center', va='center', color='white', fontweight='bold', fontsize=12)
        if group_c > 3:
            ax.text(group_a + group_b + group_c/2, 0, f"{group_c}%", ha='center', va='center', color='white', fontweight='bold', fontsize=12)

        ax.set_xlim(0, 100)
        ax.set_xlabel("Percentage (%)", fontsize=12)
        ax.set_title("Relative Frequency: 100% Stacked Bar", fontsize=15, pad=20)
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)
        ax.grid(axis='x', linestyle='--', alpha=0.5)

        # Hide y-axis since there's only one bar
        ax.get_yaxis().set_visible(False)

    plt.tight_layout()
    plt.show()

print("Adjust Group A and Group B. Group C will auto-calculate to ensure the total is always 100%.")

# Create Interactive UI Elements
slider_a = widgets.IntSlider(value=30, min=0, max=100, step=1, description='Group A %:')
slider_b = widgets.IntSlider(value=50, min=0, max=100, step=1, description='Group B %:')
chart_dropdown = widgets.Dropdown(
    options=['Pie Chart', 'Donut Chart', 'Stacked Bar Chart'],
    value='Donut Chart',
    description='Chart Type:'
)

# Group layout
ui = widgets.VBox([
    widgets.HBox([slider_a, slider_b]),
    chart_dropdown
])

# Bind logic
out = widgets.interactive_output(
    update_proportions,
    {'group_a': slider_a, 'group_b': slider_b, 'chart_type': chart_dropdown}
)

display(ui, out)
```

### What to Look For in the Dashboard

Once the code is running, experiment with the sliders and dropdowns to observe these core data principles:

* **The 100% Constraint:** Try to set Group A to 80% and Group B to 50%. The engine will actively prevent you from doing this (capping Group B at 20%). In relative frequency, a closed system can never exceed 100%.
* **The Auto-Fill Effect:** Notice how Group C actively shrinks as you grow A and B. This highlights why relative percentages can be misleading in the news: if one category's percentage goes up, another's *must* go down, even if the absolute counts for both grew.
* **Visual Weight Comparison:** Switch between the Pie Chart and the Stacked Bar Chart. Notice how it is much easier to compare the exact size difference between groups on the flat, linear scale of the Stacked Bar Chart compared to the angles of a Pie Chart.
