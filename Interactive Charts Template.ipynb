title: Plotly Interactivity Demo
---

Drag the slider. This chart is fully interactive with **no Python kernel** and no Binder — every frame is precomputed and embedded in the page.

```python
import numpy as np
import plotly.graph_objects as go

rng = np.random.default_rng(0)
n = 300
base = rng.normal(size=n)
noise = rng.normal(size=n)

# Precompute one frame per slider position. This is the key idea: the slider steps
# through frames that already exist in the figure, so the browser needs no kernel.
r_values = np.round(np.arange(-0.95, 0.96, 0.05), 2)

frames = []
for r in r_values:
    y = r * base + np.sqrt(max(0.0, 1 - r**2)) * noise
    frames.append(
        go.Frame(
            name=str(r),
            data=[go.Scatter(x=base, y=y, mode="markers",
                             marker=dict(size=7, opacity=0.7))],
            layout=go.Layout(title=f"Pearson r = {r:+.2f}"),
        )
    )

first = frames[len(frames) // 2]

fig = go.Figure(data=first.data, frames=frames, layout=first.layout)
fig.update_layout(
    template="simple_white",
    height=520,
    xaxis=dict(title="x", range=[-4, 4]),
    yaxis=dict(title="y", range=[-4, 4]),
    sliders=[dict(
        active=len(frames) // 2,
        currentvalue=dict(prefix="r = "),
        steps=[dict(method="animate", label=f.name,
                    args=[[f.name], dict(mode="immediate",
                                         frame=dict(duration=0, redraw=True))])
               for f in frames],
    )],
)
fig
```
